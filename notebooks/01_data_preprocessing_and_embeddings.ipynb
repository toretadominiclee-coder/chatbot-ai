## 4. Hugging Face Dense Embedding Generation & Emotional Retrieval Demo
We generate 384-dimensional dense vectors using Hugging Face and test retrieval on student emotional confessions and panic queries.

```python
search_demo = SemanticSearchDemo()

queries = [
    ("I feel emotionally numb and bottle up my feelings. How can I start expressing what I feel?", "Emotional Expression & Journaling"),
    ("I get intense panic attacks when the professor calls on me in class and my voice shakes.", "Social Anxiety & Classroom Coping")
]

for q, cat_filter in queries:
    print("=" * 80)
    print(f"QUERY: \"{q}\"")
    print(f"CATEGORY: {cat_filter}")
    print("-" * 80)
    results = search_demo.search(q, top_k=2, category_filter=cat_filter)
    for rank, r in enumerate(results, 1):
        print(f"[Rank #{rank}] Cosine Similarity: {r['score']} | Chunk: {r['chunk_id']}")
        print(f"Source: {r['title']} [{r['category']}]")
        print(f"Snippet: {r['snippet']}\n")
```
